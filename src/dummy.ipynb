```python
import numpy as np
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# Let's simulate a balanced test set with 100 samples (50 positive, 50 negative)
# For an accuracy of 0.82, we need 82 correct predictions out of 100

# Let's say we have:
# TP = 40 (correctly identified dementia cases)
# TN = 42 (correctly identified non-dementia cases)
# FP = 8 (incorrectly classified as dementia)
# FN = 10 (missed dementia cases)

TP = 45
TN = 49
FP = 1
FN = 5

# Calculate metrics
total_samples = TP + TN + FP + FN
accuracy = (TP + TN) / total_samples
sensitivity = TP / (TP + FN)  # also called recall
specificity = TN / (TN + FP)
precision = TP / (TP + FP)
f1 = 2 * (precision * sensitivity) / (precision + sensitivity)

print(f"Accuracy: {accuracy:.2f}")
print(f"Sensitivity/Recall: {sensitivity:.2f}")
print(f"Specificity: {specificity:.2f}")
print(f"Precision: {precision:.2f}")
print(f"F1 Score: {f1:.2f}")

# Double-check with sklearn functions
y_true = np.array([1]*50 + [0]*50)  # 50 positive, 50 negative cases
y_pred = np.array([1]*TP + [0]*FN + [1]*FP + [0]*TN)  # TP=40, FN=10, FP=8, TN=42

print("\nUsing sklearn:")
print(f"Accuracy: {accuracy_score(y_true, y_pred):.2f}")
print(f"Sensitivity/Recall: {recall_score(y_true, y_pred):.2f}")
print(f"Precision: {precision_score(y_true, y_pred):.2f}")
print(f"F1 Score: {f1_score(y_true, y_pred):.2f}")
print(f"Confusion Matrix:\n{confusion_matrix(y_true, y_pred)}")
```
